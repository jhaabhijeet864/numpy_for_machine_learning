# Advanced Indexing

```python
import numpy as np 
```

## Fancy Indexing

```python
arr = np.array(['a', 'b', 'c', 'd', 'e', 'f'])
index_list = np.array([0, 4, 1, 4])

selection = arr[index_list]
print(f"Selection: {selection}")
```

```python
matrix = np.array([[10, 20, 30],
                   [40, 50, 60],
                   [70, 80, 90]])

row_indices = np.array([0, 2, 1])
col_indices = np.array([0, 1, 2])

selection_pairs = matrix[row_indices, col_indices]
print(f"Selection Pairs: {selection_pairs}")
```

## Boolean Indexing

Boolean indexing is powerful because it allows to filter data based on complex condition.

#### Mask Creation 

Mechanism: You create a boolean array (mask) of the same shape as the data array, where True indicates the element to keep.

Goal: Select all elements that satisfy a condition, often used for data cleaning or feature engineering.

```python
data = np.random.randint(0, 100, size=(4,4))
print(f"Original Data:\n {data}")
```

```python
# Create a boolean mask: True for values greater than 50
mask = data > 60
print(f"\n Boolean Mask:\n {mask}")
```

```python
# Apply the mask
high_values = data[mask]
print(f"\n High Values (1D Object): {high_values}")
```
