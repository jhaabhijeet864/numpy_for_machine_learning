# Linear Algebra Operations

Numpy's sub-module "numpy.linlg", provides set of tools for performing linear algebra computations effectivly.

```python
import numpy as np
```

```python
A = np.array([[1,2],[3,4]])
B = np.array([[5,6],[7,8]])
vector = np.array([1,0])
```

```python
# 1. Matrix multiplication 
C = A @ B
print("A @ B (Matrix Product):\n", C)
```

```python
# 2. Matrix Vector Product
result = A @ vector
print("\nA @ vector:\n", result)
```

##### The numpy.linalg module offers many essential functions for working with matrices:

```python
# Finding the Inverse Matrix
N = np.array([[4,2], [1,3]])

# Calculating the Inverse Matrix
N_inv = np.linalg.inv(N)
print("Inverse of N:\n", N_inv)

# Verification: m * m_inv should be identity matrix
identity = N * N_inv
print("\n Verification (M * M_inv)::\n", identity.round(decimals=5))

```
