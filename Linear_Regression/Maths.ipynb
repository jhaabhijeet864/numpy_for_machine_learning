### 1. Math (normal equation)

This normal equation, directly calculates the optimal weights ($\theta$) using linear algebra.

Where:

$X$: The design matrix (features + a column of ones for the intercept/bias).

$y$: The target vector.

$X^T$: The transpose of $X$.

$(X^T X)^{-1}$: The inverse of the matrix product $X^T X$.

$\theta$: The calculated weight vector (coefficients and intercept).
