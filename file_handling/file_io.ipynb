# 💾 Python File Handling: Opening and Closing

Three main interacting main steps:
1. Opening the file
2. Performing operations (read/write)
3. Closing the file

```python
import numpy as np
```

## open() functions

file_object = open(filename, mode)

The with statement is the best practice in Python. It automatically handles closing the file, even if errors occur.

```python
with open('hello_world.txt', 'r') as file:
    # Operations (read/write) are performed inside this block
    file_content = file.read()
print("file_content:", file_content)

# Once Python exits the 'with' block, the file is automatically closed.
```

The os file part assings the opened file object to the variable "file" for use within the block

## Write() and append() functions

### .write()

```python
data_to_write = "Hello, World!\n"
data_to_write += "This is the first line of data for my AIML Project."

# Caution: If hello_world.txt exists, its content is ERASED.
with open("hello_world.txt", 'w') as f:
    f.write(data_to_write)
```
